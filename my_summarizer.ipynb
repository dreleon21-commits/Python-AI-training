```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

name = "sshleifer/distilbart-cnn-12-6"
tokenizer = AutoTokenizer.from_pretrained(name)
model = AutoModelForSeq2SeqLM.from_pretrained(name)

def summarizer(text, max_length=80, min_length=25, do_sample=False, truncation=True):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=1024)
    ids = model.generate(**inputs, max_length=max_length,
                         min_length=min_length, do_sample=do_sample)
    return [{"summary_text": tokenizer.decode(ids[0], skip_special_tokens=True)}]

print("Model is ready!")
```

```python
def chunk_text(text, max_words=400):
    words = text.split()
    return [" ".join(words[i:i + max_words])
            for i in range(0, len(words), max_words)]

def summarize(text):
    text = text.strip()
    if len(text.split()) < 40:
        return "Text is too short. Please paste at least a few sentences."

    pieces = []
    for chunk in chunk_text(text):
        out = summarizer(chunk, max_length=80, min_length=25,
                         do_sample=False, truncation=True)
        pieces.append(out[0]["summary_text"])

    if len(pieces) == 1:
        return pieces[0]

    combined = " ".join(pieces)
    final = summarizer(combined, max_length=80, min_length=25,
                       do_sample=False, truncation=True)
    return final[0]["summary_text"]

print("Function ready!")
```

```python
article = """Coffee is one of the most widely consumed drinks in the world, and its history stretches back many centuries. According to legend, it was first discovered in Ethiopia when a goat herder noticed that his goats became unusually energetic after eating berries from a certain tree. From there, coffee spread across the Arabian Peninsula, where it became popular in trading cities and social gatherings. By the seventeenth century it had reached Europe, where coffee houses became places for conversation, business, and the exchange of new ideas. Today, coffee is grown in dozens of countries across Latin America, Africa, and Asia, and millions of farmers depend on it for their income. Researchers continue to study its effects on health, and many studies suggest that moderate consumption is safe for most adults."""

print(summarize(article))
```

```python
import gradio as gr

demo = gr.Interface(
    fn=summarize,
    inputs=gr.Textbox(lines=12, label="Paste your article"),
    outputs=gr.Textbox(label="Summary"),
    title="My Text Summarizer",
    description="Paste a long article and get a short summary.",
)

demo.launch()
```
